```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df['Review Rating'].isna().sum()
```

```python
df.columns
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df.rename(columns={"purchase_amount_(usd)":"purchase_amount"},inplace=True)
```

```python
lables=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=lables)
```

```python
df['age_group'].value_counts()
```

```python
df['frequency_of_purchases'].unique()
```

```python
freq_map={
    'Fortnightly':14,
    'Weekly':7,
    'Annually':365,
    'Quarterly': 90,
    'Bi-Weekly':14,
    'Monthly':30, 
    'Every 3 Months':90
}
df['purchases_frequency']=df['frequency_of_purchases'].map(freq_map)
```

```python
from getpass import getpass
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

MYSQL_HOST = "localhost"
MYSQL_PORT = 3306
MYSQL_USER = "root"
MYSQL_PASSWORD = getpass("Enter your MySQL password: ")
MYSQL_DATABASE = "customer_analysis"

# Connect to the MySQL server first (without selecting a database).
server_url = URL.create(
    "mysql+pymysql",
    username=MYSQL_USER,
    password=MYSQL_PASSWORD,
    host=MYSQL_HOST,
    port=MYSQL_PORT,
)

server_engine = create_engine(server_url)

# Verify the server connection and create the database if necessary.
with server_engine.connect() as connection:
    print("Connected to MySQL server:", connection.execute(text("SELECT VERSION()")).scalar())
    connection.execute(text(f"CREATE DATABASE IF NOT EXISTS `{MYSQL_DATABASE}`"))
    connection.commit()

# Connect to the project database.
database_url = URL.create(
    "mysql+pymysql",
    username=MYSQL_USER,
    password=MYSQL_PASSWORD,
    host=MYSQL_HOST,
    port=MYSQL_PORT,
    database=MYSQL_DATABASE,
)

engine = create_engine(database_url)
with engine.connect() as connection:
    print("Connected to database:", connection.execute(text("SELECT DATABASE()")).scalar())
```

```python
# Check that the DataFrame exists and preview its shape.
print("DataFrame shape:", df.shape)
display(df.head())

# Upload the cleaned data to MySQL.
df.to_sql(
    name="customer_shopping_behavior_clean",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=1000,
)

print("Upload complete.")
```
